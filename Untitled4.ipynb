```python
import pandas as pd
from sqlalchemy import create_engine

# Define the connection string (user:password@host:port/database)
engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')

# Test the connection
try:
    engine.connect()
    print("Successfully connected to the database!")
except Exception as e:
    print(f"Error connecting to the database: {e}")
```

```python
from tqdm.auto import tqdm

# Define the table name and the URL for the full dataset
table_name = 'yellow_taxi_data'
prefix = 'https://github.com/DataTalksClub/nyc-tlc-data/releases/download/yellow/'

# Define data types to prevent DtypeWarnings
dtype = {
    "VendorID": "Int64",
    "passenger_count": "Int64",
    "trip_distance": "float64",
    "RatecodeID": "Int64",
    "store_and_fwd_flag": "string",
    "PULocationID": "Int64",
    "DOLocationID": "Int64",
    "payment_type": "Int64",
    "fare_amount": "float64",
    "extra": "float64",
    "mta_tax": "float64",
    "tip_amount": "float64",
    "tolls_amount": "float64",
    "improvement_surcharge": "float64",
    "total_amount": "float64",
    "congestion_surcharge": "float64"
}

parse_dates = ["tpep_pickup_datetime", "tpep_dropoff_datetime"]

print("Starting data ingestion...")

# 1. Read the CSV in chunks (iterator=True)
df_iter = pd.read_csv(
    prefix + 'yellow_tripdata_2021-01.csv.gz',
    dtype=dtype,
    parse_dates=parse_dates,
    iterator=True,
    chunksize=100000
)

# 2. Get the first chunk to establish the table schema
first_chunk = next(df_iter)

# 3. Create the empty table in Postgres
first_chunk.head(0).to_sql(name=table_name, con=engine, if_exists='replace')

# 4. Insert the first chunk
first_chunk.to_sql(name=table_name, con=engine, if_exists='append')

# 5. Loop through the remaining chunks and insert them with a progress bar
for chunk in tqdm(df_iter):
    chunk.to_sql(name=table_name, con=engine, if_exists='append')

print("Data ingestion complete!")
```
